```python
from __future__ import annotations

import argparse
import logging
from pathlib import Path
from typing import Any

from pydantic import TypeAdapter

from pipeline.builder.book_builder import build_book_manifests
from pipeline.builder.final_json_builder import seed_final_questions
from pipeline.docling.extract_book import extract_book_with_docling
from pipeline.docling.import_existing import import_existing_docling_output
from pipeline.models import ChapterManifest, Question
from pipeline.parser.answer_parser import parse_answers_from_chapter
from pipeline.parser.chapter_index import load_chapter_index
from pipeline.parser.chapter_splitter import split_chapters_by_markers
from pipeline.parser.matcher import match_answers_to_questions
from pipeline.parser.question_parser import parse_questions_from_chapter
from pipeline.storage.json_io import (
    read_json,
    read_model,
    write_json_atomic,
    write_text_atomic,
)
from pipeline.storage.paths import PipelinePaths, default_paths
from pipeline.validator.question_validator import validate_questions_locally
from pipeline.validator.repair_detector import build_all_repair_batches


LOGGER = logging.getLogger(__name__)


def parse_args() -> argparse.Namespace:
    parser = argparse.ArgumentParser(
        description="Run the local OCR/question pipeline for one book PDF."
    )
    parser.add_argument("--pdf", type=Path, required=True)
    parser.add_argument("--subject", required=True)
    parser.add_argument("--book-id", required=True)
    parser.add_argument("--title", required=True)
    parser.add_argument("--chapter-index", type=Path, required=True)
    parser.add_argument(
        "--existing-docling-output",
        type=Path,
        default=None,
        help="Use an existing Docling output folder instead of running Docling.",
    )
    parser.add_argument("--start-page", type=int, default=None)
    parser.add_argument("--end-page", type=int, default=None)
    parser.add_argument("--batch-size", type=int, default=5)
    parser.add_argument("--replace-existing", action="store_true")
    return parser.parse_args()


def process_book(
    *,
    pdf: Path,
    subject: str,
    book_id: str,
    title: str,
    chapter_index: Path,
    existing_docling_output: Path | None = None,
    start_page: int | None = None,
    end_page: int | None = None,
    batch_size: int = 5,
    replace_existing: bool = False,
    paths: PipelinePaths | None = None,
) -> dict[str, Any]:
    """Run the full local pipeline up to repair input preparation."""

    paths = paths or default_paths()
    chapter_entries = load_chapter_index(chapter_index)
    page_start = start_page or min(entry.pdf_pages.start for entry in chapter_entries)
    page_end = end_page or max(entry.pdf_pages.end for entry in chapter_entries)

    if existing_docling_output is not None:
        import_existing_docling_output(
            book_id=book_id,
            source_pdf=pdf,
            existing_output_dir=existing_docling_output,
            paths=paths,
            replace_existing=replace_existing,
        )
    else:
        extract_book_with_docling(
            book_id=book_id,
            source_pdf=pdf,
            page_start=page_start,
            page_end=page_end,
            paths=paths,
            replace_existing=replace_existing,
        )

    book_manifest = build_book_manifests(
        book_id=book_id,
        subject=subject,
        title=title,
        source_file=Path(pdf).name,
        chapter_index_path=chapter_index,
        paths=paths,
        replace_existing=replace_existing,
        update_subject=True,
    )

    markdown_text = paths.docling_markdown_path(book_id).read_text(encoding="utf-8")
    chapter_markdown = split_chapters_by_markers(markdown_text, chapter_entries)

    summary: dict[str, Any] = {
        "book_id": book_id,
        "chapters": [],
    }

    question_adapter = TypeAdapter(list[Question])
    for entry in chapter_entries:
        manifest = read_model(
            paths.chapter_manifest_path(book_id, entry.chapter_number),
            ChapterManifest,
        )
        chapter_text = chapter_markdown[entry.chapter_number]
        chapter_dir = paths.chapter_dir(book_id, entry.chapter_number)

        questions_raw = parse_questions_from_chapter(
            markdown_text=chapter_text,
            book_id=manifest.book_id,
            chapter_id=manifest.chapter_id,
            subject=manifest.subject,
        )
        write_text_atomic(
            chapter_dir / "chapter.md",
            chapter_text,
            overwrite=replace_existing,
        )
        write_json_atomic(
            paths.questions_raw_path(book_id, entry.chapter_number),
            {"questions": questions_raw},
            overwrite=replace_existing,
        )

        answers = parse_answers_from_chapter(
            markdown_text=chapter_text,
            chapter_id=manifest.chapter_id,
        )
        matched = match_answers_to_questions(questions_raw, answers)
        validated = validate_questions_locally(matched)
        write_json_atomic(
            paths.questions_validated_path(book_id, entry.chapter_number),
            {
                "chapter_id": manifest.chapter_id,
                "answers_detected": answers,
                "questions": validated,
            },
            overwrite=replace_existing,
        )

        # Re-validate through the JSON contract before preparing repair batches.
        validated_payload = read_json(
            paths.questions_validated_path(book_id, entry.chapter_number)
        )
        validated_questions = question_adapter.validate_python(
            validated_payload["questions"]
        )
        batches = build_all_repair_batches(
            chapter_title=manifest.title,
            questions=validated_questions,
            batch_size=batch_size,
            chapter_markdown_path=chapter_dir / "chapter.md",
        )
        final_questions = seed_final_questions(validated_questions)
        write_json_atomic(
            paths.repair_input_path(book_id, entry.chapter_number),
            {
                "chapter_id": manifest.chapter_id,
                "batch_size": batch_size,
                "total_batches": len(batches),
                "total_questions_for_repair": sum(
                    len(batch.questions) for batch in batches
                ),
                "batches": batches,
            },
            overwrite=replace_existing,
        )
        write_json_atomic(
            paths.questions_final_path(book_id, entry.chapter_number),
            {
                "chapter_id": manifest.chapter_id,
                "questions": final_questions,
            },
            overwrite=replace_existing,
        )

        chapter_summary = {
            "chapter_number": entry.chapter_number,
            "subject_chapter_number": entry.subject_chapter_number
            or entry.chapter_number,
            "title": entry.title,
            "questions": len(validated_questions),
            "answers_detected": len(answers),
            "repair_batches": len(batches),
            "questions_for_repair": sum(len(batch.questions) for batch in batches),
        }
        summary["chapters"].append(chapter_summary)
        LOGGER.info(
            "Processed chapter %s: %s questions, %s repair batches",
            entry.chapter_number,
            chapter_summary["questions"],
            chapter_summary["repair_batches"],
        )

    summary["total_chapters"] = len(book_manifest.chapters)
    return summary


def main() -> None:
    logging.basicConfig(
        level=logging.INFO,
        format="%(asctime)s | %(levelname)s | %(message)s",
    )
    args = parse_args()
    summary = process_book(
        pdf=args.pdf,
        subject=args.subject,
        book_id=args.book_id,
        title=args.title,
        chapter_index=args.chapter_index,
        existing_docling_output=args.existing_docling_output,
        start_page=args.start_page,
        end_page=args.end_page,
        batch_size=args.batch_size,
        replace_existing=args.replace_existing,
    )
    LOGGER.info("Processed book %s with %s chapters", args.book_id, summary["total_chapters"])


if __name__ == "__main__":
    main()
```

```python
%%writefile /content/pipeline/models/__init__.py
from pydantic import BaseModel
from typing import Optional, List
from pathlib import Path

class ChapterManifest(BaseModel):
    # Minimal definition inferred from process_book.py
    book_id: str
    chapter_id: str
    subject: str
    title: str

class Question(BaseModel):
    # Minimal definition inferred from process_book.py
    question_id: str
    text: str
    # Add other common fields for Question if available or needed

class BookChapterSummary(BaseModel):
    # Minimal definition inferred from process_book.py
    chapter_number: int
    subject_chapter_number: Optional[int] = None
    title: str
    questions: int
    answers_detected: int
    repair_batches: int
    questions_for_repair: int

class BookManifest(BaseModel):
    # Minimal definition inferred from process_book.py
    book_id: str
    subject: str
    title: str
    source_file: str
    chapter_index_path: Path
    chapters: List[ChapterManifest] = [] # Assuming it will contain chapter manifests
```
